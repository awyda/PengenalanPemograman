```python
nama_pelanggan = input("Masukkan Nama Pelanggan: ")
total_belanja = float(input("Masukkan Total Belanja Sebelum Diskon: "))
persen_diskon = float(input("Masukkan Persentase Diskon (contoh: 12.5): "))

nominal_diskon = total_belanja * (persen_diskon / 100)
total_bayar = total_belanja - nominal_diskon

str_total_awal = f"Rp{total_belanja:,.2f}"
str_diskon = f"Rp{nominal_diskon:,.2f}"
str_total_bayar = f"Rp{total_bayar:,.2f}"

print("\n" + "=" * 36)
print(f"{'NOTA PEMBAYARAN':^36}")
print(f"{f'Pelanggan: {nama_pelanggan}':^36}")
print("=" * 36)

print(f"{'Total Awal':<18}{str_total_awal:>18}")
print(f"{f'Diskon ({persen_diskon}%)':<18}{str_diskon:>18}")
print("-" * 36)
print(f"{'Total Bayar':<18}{str_total_bayar:>18}")
print("=" * 36)
```
